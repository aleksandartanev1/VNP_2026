```python
# ==========================================
# ВЕЖБА 1: Променливи и пресметки
# Задача: Пресметај плоштина и периметар на правоаголник.
# ==========================================

sirina = 5
visina = 10

plostina = sirina * visina
perimetar = 2 * (sirina + visina)

print("--- Вежба 1 ---")
print(f"Плоштина: {plostina}")
print(f"Периметар: {perimetar}\n")
```

```python
# ==========================================
# ВЕЖБА 2: Креирање DataFrame и основен преглед
# Во оваа вежба ќе креираме табела со податоци за продажба на производи и ќе ги видиме основните информации.
# ==========================================

import pandas as pd

# Податоци за продажба
data = {
    'Продавач': ['Александар', 'Марија', 'Александар', 'Игор', 'Марија', 'Игор'],
    'Производ': ['Лаптоп', 'Мравка (Глувче)', 'Монитор', 'Лаптоп', 'Тастатура', 'Монитор'],
    'Цена': [45000, 1200, 8500, 45000, 2500, 8500],
    'Количина': [1, 3, 2, 2, 4, 1]
}

# Креирање на DataFrame
df = pd.DataFrame(data)
print("--- Вежба 2 ---")
print("--- Основни информации за DataFrame ---")
display(df.head())  # Ги прикажува првите редови (или print(df) ако не користиш Jupyter)
print("\nИнформации за типовите на податоци:")
df.info()
```

```python
# ==========================================
# ВЕЖБА 3: Филтрирање и селектирање податоци
#Како да издвоиш редови што исполнуваат одредени услови (на пр. продажба над одредена сума или од конкретен продавач).
# ==========================================


# Пресметување вкупна вредност за секоја продажба
df['Вкупно'] = df['Цена'] * df['Количина']

print("--- Сите продажби со пресметана вкупна сума ---")
display(df)

# Филтер 1: Продажби каде продавач е Марија
marija_prodazbi = df[df['Продавач'] == 'Марија']
print("\n--- Продажби направени од Марија ---")
display(marija_prodazbi)

# Филтер 2: Продажби со вкупна вредност поголема од 10,000 денари
skapi_prodazbi = df[df['Вкупно'] > 10000]
print("\n--- Продажби со вредност > 10,000 денари ---")
display(skapi_prodazbi)
```

```python
# ==========================================
# ВЕЖБА 4: Групирање и агрегација (groupby)
#Многу моќна алатка за пресметување суми, просеци или статистички податоци по категории.
# ==========================================
# Колку вкупно пари остварил секој продавач?
zarabotka_po_prodavac = df.groupby('Продавач')['Вкупно'].sum().reset_index()

print("--- Вкупна заработка по продавач ---")
display(zarabotka_po_prodavac)

# Колку парчиња од секој производ се продадени вкупно?
kolicina_po_proizvod = df.groupby('Производ')['Количина'].sum().reset_index()

print("\n--- Продадена количина по вид на производ ---")
display(kolicina_po_proizvod)
```

```python
# ==========================================
# ВЕЖБА 5:Ракување со недостасувачки податоци (NaN) и сортирање
#Во реалните проекти често има празни места (Null/NaN) кои треба да се исчистат
# ==========================================
import numpy as np

# Додавање податоци со намерно празни места (NaN)
data_so_dupki = {
    'Продавач': ['Александар', 'Марија', 'Игор', 'Александар'],
    'Регион': ['Скопје', np.nan, 'Битола', 'Скопје'],
    'Оцена': [4.5, 3.8, np.nan, 24.9]
}

df_dirty = pd.DataFrame(data_so_dupki)
print("--- Оригинални податоци со NaN вредности ---")
display(df_dirty)

# Пополнување на празни места со стандардна вредност (на пр. 'Непознато' или просек)
df_clean = df_dirty.copy()
df_clean['Регион'] = df_clean['Регион'].fillna('Непознат регион')
df_clean['Оцена'] = df_clean['Оцена'].fillna(df_dirty['Оцена'].mean())

print("\n--- Исчистени податоци ---")
display(df_clean)

# Сортирање на првичниот df по Вкупно опаѓачки (од најголема кон најмала продажба)
df_sorted = df.sort_values(by='Вкупно', ascending=False)
print("\n--- Сортирани продажби по вредност (опаѓачки) ---")
display(df_sorted)
```
